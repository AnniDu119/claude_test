```python
class Animal:
    """Base class demonstrating inheritance and encapsulation"""
    
    def __init__(self, name, species):
        self.name = name
        self.species = species
        self._age = 0  # Protected attribute
    
    def speak(self):
        """Method to be overridden by subclasses"""
        return f"{self.name} makes a sound"
    
    def have_birthday(self):
        """Encapsulation: control internal state"""
        self._age += 1
        return f"{self.name} is now {self._age} years old"
```

```python
dog = Animal('Wolfie','chiwawa')
```

```python
dog.have_birthday()
```

```python
dog.speak()
```

```python
cat = Animal('mochi','british shorthair')
```
